```python
import pandas as pd
import numpy as np
import random


n_rows = 10000
data = []

for i in range(n_rows):
    income = random.randint(300000, 5000000)
    invoice_amount = random.randint(10000, 1500000)
    dpd = random.randint(0, 45)
    sector = random.choice(['Gıda', 'İnşaat', 'Tekstil', 'Sağlık', 'Metal'])
    
  
    logic_score = 70 
    
   
    if dpd > 20: logic_score -= 40
    if invoice_amount > income * 0.3: logic_score -= 30
    if sector == 'İnşaat': logic_score -= 20
    if sector == 'Gıda': logic_score += 15
    
   
    if logic_score < 50:
        ptp_status = 1 if random.random() < 0.15 else 0 
    else:
        ptp_status = 1 if random.random() < 0.85 else 0 

    data.append({
        'customer_id': f"CUST-{i}",
        'income': income,
        'invoice_amount': invoice_amount,
        'dpd': dpd,
        'sector': sector,
        'tax_type': random.choice(['STANDART', 'TEVKİFAT']),
        'factoring_type': random.choice(['With Recourse', 'Non-Recourse']),
        'contact_type': random.choice(['CALL', 'SMS']),
        'ptp_status': ptp_status
    })

df_v4 = pd.DataFrame(data)
df_v4.to_csv(r"C:\Users\yildi\Desktop\tamfinans\collection_master_data_v4.csv", index=False, encoding='utf-8-sig')
```
