```python
import pandas as pd
import numpy as np
import random


n_rows = 10000
sectors = ['Perakende', 'Enerji', 'Metal', 'Otomotiv', 'İnşaat', 'Sağlık', 'Gıda', 'Tekstil']
sector_risks = {'Perakende': 0.1, 'Enerji': 0.05, 'Metal': 0.15, 'Otomotiv': 0.1, 
                'İnşaat': 0.4, 'Sağlık': 0.05, 'Gıda': 0.05, 'Tekstil': 0.3}

data = []
```

```python
for i in range(n_rows):
    customer_id = f"CUST-{1000 + i}"
    income = random.randint(300000, 5000000)
    sector = random.choice(sectors)
    invoice_amount = random.randint(10000, 1500000)
    dpd = random.randint(0, 45) 
    
  
    prob = 0.90
    
    if dpd > 20: prob -= 0.4
    if dpd > 35: prob -= 0.3
    
    prob -= sector_risks[sector]
    
    if invoice_amount / income > 0.3:
        prob -= 0.2
    
    ptp_status = 1 if random.random() < prob else 0
    
    data.append({
        'customer_id': customer_id,
        'income': income,
        'sector': sector,
        'legal_type': random.choice(['Corporate (VKN)', 'Individual (TCKN)']),
        'invoice_amount': invoice_amount,
        'invoice_date': '01.01.2026', 
        'due_date': '15.01.2026',
        'dpd': dpd,
        'tax_type': random.choice(['STANDART', 'TEVKİFAT']),
        'factoring_type': random.choice(['With Recourse', 'Non-Recourse']),
        'contact_type': random.choice(['CALL', 'SMS']),
        'ptp_status': ptp_status
    })
```

```python
df_v3 = pd.DataFrame(data)
file_path = r"C:\Users\yildi\Desktop\tamfinans\collection_master_data_v3.csv"
df_v3.to_csv(file_path, index=False, encoding='utf-8-sig')

print(f"Başarılı! 10.000 satırlık  veri seti v3 olarak kaydedildi: {file_path}")
print("PTP Oranı:\n", df_v3['ptp_status'].value_counts(normalize=True))
```
